# MNIST Handwritten Digit Recognition - Exploratory Data Analysis & Model Experiments

This notebook demonstrates exploratory data analysis (EDA), dataset visualization, preprocessing transformations, CNN model architecture, and Grad-CAM explainability for the Handwritten Digit Recognition system.

```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

# Append project root
BASE_DIR = Path(".").resolve().parent
sys.path.append(str(BASE_DIR))

from src.data_loader import MNISTDataLoader
from src.preprocess import ImagePreprocessor
from src.model import DigitCNNModel
from src.predict import DigitPredictor

print(f"TensorFlow Version: {tf.__version__}")
```

## 1. Load Dataset & Summary Statistics

```python
loader = MNISTDataLoader()
(X_train_raw, y_train), (X_val_raw, y_val), (X_test_raw, y_test) = loader.load_data()
summary = loader.get_dataset_summary()
print("Dataset Summary:", summary)
```

## 2. Visualize Sample Digits & Class Distribution

```python
plt.figure(figsize=(10, 4))
unique, counts = np.unique(y_train, return_counts=True)
sns.barplot(x=unique, y=counts, palette="viridis")
plt.title("MNIST Training Class Distribution", fontsize=14, fontweight="bold")
plt.xlabel("Digit Class", fontsize=12)
plt.ylabel("Sample Count", fontsize=12)
plt.show()
```

## 3. Preprocessing Verification

```python
preprocessor = ImagePreprocessor()
X_train_proc = preprocessor.preprocess_dataset(X_train_raw)
print(f"Processed Training Tensor Shape: {X_train_proc.shape}, Min: {X_train_proc.min()}, Max: {X_train_proc.max()}")
```

## 4. Build & Inspect CNN Architecture
